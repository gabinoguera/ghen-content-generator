## 1. CONFIGURACIÓN INICIAL


```python
# === IMPORTACIÓN UNIFICADA DEL MÓDULO ===
# Celda única para cargar todo el sistema

import importlib
import sys

# Recargar módulo si ya está cargado
if 'longcontent_generator' in sys.modules:
    importlib.reload(sys.modules['longcontent_generator'])

# Importar todo
from longcontent_generator import *

print("✅ Módulo LongContent Generator v1.0.0 cargado correctamente")
print("🔧 Pipeline simplificado activo")
print("📋 Funciones principales disponibles:")
print("   • google_custom_search() - Búsqueda en Google")
print("   • scrape_articles_batch() - Scraping de artículos")
print("   • analyze_articles_batch() - Análisis SEO con Gemini")
print("   • generate_article_from_outline() - Generación de contenido")
print("   • qa_article_coverage() - Análisis de calidad")
print("   • publish_article_from_markdown_cleaned() - Publicación WordPress")
print("   • extract_keywords_from_search_titles() - Keywords de títulos ✅")
print("   • load_seo_keywords_from_analysis() - Keywords del análisis ✅")
```

```python
# === CONFIGURACIÓN DEL PIPELINE ===
# Configuración del pipeline (SIMPLIFICADO)
MAIN_QUERY = "como escribir una novela"  # Cambia aquí tu keyword principal
LANGUAGE = "es"
COUNTRY = "es"

print(f"🎯 CONFIGURACIÓN SIMPLIFICADA:")
print(f"   📌 Keyword principal: {MAIN_QUERY}")
print(f"   🌍 País/Idioma: {COUNTRY}/{LANGUAGE}")
print(f"\n💡 Para cambiar la keyword, modifica MAIN_QUERY arriba")
```

## 🔍 PASO 1: BÚSQUEDA EN GOOGLE (COMPETENCIA)


```python
# === PASO 1: BÚSQUEDA EN GOOGLE ===
print("🔍 PASO 1: Buscando competencia en Google...")
print("="*50)

# Buscar competencia
search_results_df = google_custom_search(MAIN_QUERY)

if not search_results_df.empty:
    print(f"\n✅ {len(search_results_df)} resultados encontrados")
    print("\n📋 Títulos encontrados:")
    for i, row in search_results_df.iterrows():
        print(f"   {i+1}. {row['title'][:80]}...")
    
    # Guardar resultados
    search_results_df.to_csv('search_results.csv', index=False)
    print(f"\n💾 Resultados guardados en: search_results.csv")
else:
    print("❌ No se encontraron resultados")
```

## 3. 📄 PASO 2: SCRAPING DE ARTÍCULOS


```python
print("📄 PASO 2: Scrapeando artículos de la competencia...")
print("="*50)

if 'search_results_df' in locals() and not search_results_df.empty:
    # Scrapear artículos
    scraped_df = scrape_articles_batch(search_results_df)
    
    # Guardar resultados
    scraped_df.to_csv('scraped_articles.csv', index=False)
    print(f"\n💾 Artículos scrapeados guardados en: scraped_articles.csv")
    print(f"\n📊 Resumen del scraping:")
    
    successful_scrapes = scraped_df['scraped_text'].str.strip().astype(bool).sum()
    print(f"   ✅ Exitosos: {successful_scrapes}/{len(scraped_df)}")
    
    # Mostrar ejemplos de contenido extraído
    for i, row in scraped_df.iterrows():
        if row['scraped_text'].strip():
            preview = row['scraped_text'][:200].replace('\n', ' ')
            print(f"\n   📝 Artículo {i+1}: {preview}...")
else:
    print("❌ No hay resultados de búsqueda para scrapear")
    print("   Ejecuta primero la celda de búsqueda en Google")
```

## 4. 🤖 PASO 3: ANÁLISIS SEO CON GEMINI


```python
print("🤖 PASO 3: Analizando contenido con Gemini...")
print("="*50)

if 'scraped_df' in locals() and not scraped_df.empty:
    # Analizar con Gemini
    analyzed_df = analyze_articles_batch(scraped_df)
    
    # Guardar análisis
    analyzed_df.to_csv('SEO_Analysis_Results.csv', index=False)
    print(f"\n💾 Análisis SEO guardado en: SEO_Analysis_Results.csv")
    
    print(f"\n📊 Resumen del análisis:")
    for i, row in analyzed_df.iterrows():
        if 'SEO Analysis' in row and row['SEO Analysis']:
            analysis_preview = str(row['SEO Analysis'])[:300].replace('\n', ' ')
            print(f"\n   🧠 Análisis {i+1}: {analysis_preview}...")
else:
    print("❌ No hay artículos scrapeados para analizar")
    print("   Ejecuta primero las celdas de búsqueda y scraping")
```

## 5. 📝 PASO 4: GENERACIÓN DE CONTENIDO


```python
print("📝 PASO 4: Generando contenido long-form (ENFOQUE SIMPLIFICADO)...")
print("="*50)

if 'analyzed_df' in locals() and not analyzed_df.empty:
    try:
        # Cargar keywords SEO del análisis
        print("\n📥 Cargando keywords del análisis SEO...")
        related_keywords = load_seo_keywords_from_analysis('SEO_Analysis_Results.csv')
        
        if related_keywords and related_keywords.get('keywords'):
            print(f"✅ Keywords SEO cargadas: {len(related_keywords['keywords'])}")
            
            # Extraer keywords de los títulos de resultados de búsqueda
            if 'search_results_df' in locals() and not search_results_df.empty:
                title_keywords = extract_keywords_from_search_titles(search_results_df)
                print(f"✅ Keywords de títulos extraídas: {len(title_keywords)}")
                print(f"   Ejemplos: {title_keywords[:8]}")
                
                # Combinar keywords (SEO + títulos)
                all_keywords = related_keywords['keywords'] + title_keywords
                # Deduplicar
                all_keywords = list(set(all_keywords))
                print(f"✅ Keywords totales combinadas: {len(all_keywords)}")
            else:
                all_keywords = related_keywords['keywords']
            
            # Crear contexto de keywords combinado
            combined_keywords = {
                'keywords': all_keywords,
                'questions': related_keywords.get('questions', [])
            }
            
            # Análisis combinado
            print("\n🧠 Generando análisis combinado...")
            combined_analysis = analyze_combined_context(analyzed_df, combined_keywords, MAIN_QUERY)
            
            # Plan de contenido
            print("\n📋 Creando plan de contenido...")
            content_plan = create_intelligent_content_plan(combined_analysis, MAIN_QUERY, combined_keywords)
            
            # Esquema del artículo
            print("\n📝 Generando esquema del artículo...")
            outline = generate_article_outline(
                main_query=MAIN_QUERY,
                search_results=analyzed_df.to_dict('records'),
                related_keywords_context=combined_keywords,
                content_plan=content_plan
            )
            
            # Guardar esquema
            with open('esquema_articulo.md', 'w', encoding='utf-8') as f:
                f.write(outline)
            print(f"\n💾 Esquema guardado en: esquema_articulo.md")
            
            # Generar artículo completo
            print("\n✍️ Generando artículo completo...")
            article = generate_article_from_outline(
                outline_md=outline,
                content_plan=content_plan,
                related_keywords_context=combined_keywords
            )
            
            # Guardar artículo
            with open('articulo_completo.md', 'w', encoding='utf-8') as f:
                f.write(article)
            
            print(f"\n✅ ARTÍCULO GENERADO EXITOSAMENTE")
            print(f"📊 Longitud: {len(article):,} caracteres")
            print(f"📊 Palabras: {len(article.split()):,} palabras")
            print(f"\n💾 Artículo guardado en: articulo_completo.md")
            
        else:
            print("❌ No se pudieron cargar keywords del análisis")
            
    except Exception as e:
        print(f"❌ Error en la generación: {str(e)}")
        import traceback
        traceback.print_exc()
        
else:
    print("❌ No hay análisis SEO disponible")
    print("   Ejecuta primero las celdas anteriores")
```

## 6. 🔍 PASO 5: ANÁLISIS QA DEL CONTENIDO


```python
print("🔍 PASO 5: Ejecutando análisis QA del artículo...")
print("="*50)

if 'related_keywords' in locals() and 'article' in locals():
    try:
        # QA del artículo
        qa_report = qa_article_coverage(article, related_keywords, MAIN_QUERY)
        
        print("\n✅ REPORTE QA GENERADO")
        print("\n📋 Vista previa del reporte QA:")
        print("="*50)
        print(qa_report[:1000])
        if len(qa_report) > 1000:
            print(f"\n... ({len(qa_report) - 1000} caracteres más)")
        
        # Sugerencias de artículos relacionados
        print("\n💡 Generando sugerencias de artículos relacionados...")
        suggestions = suggest_related_articles(qa_report, related_keywords, MAIN_QUERY)
        
        print("\n✅ SUGERENCIAS GENERADAS")
        print("\n📋 Vista previa de sugerencias:")
        print("="*50)
        print(suggestions[:800])
        if len(suggestions) > 800:
            print(f"\n... ({len(suggestions) - 800} caracteres más)")
            
    except Exception as e:
        print(f"❌ Error en QA: {str(e)}")
        
else:
    print("❌ No hay artículo generado para analizar")
    print("   Ejecuta primero la celda de generación de contenido")
```

## 8. 📰 PASO 6: PUBLICACIÓN EN WORDPRESS (CONTROL MANUAL)

**⚠️ IMPORTANTE:** Revisa el contenido generado antes de publicar. Puedes modificar el título y el estado (draft/publish) según necesites.


```python
# CONFIGURACIÓN DE PUBLICACIÓN
ARTICLE_TITLE = "Aprende a Escribir un Libro: La Guía Definitiva para el Autor Creativo"  # Cambia el título aquí
PUBLICATION_STATUS = "draft"  # "draft" para borrador, "publish" para publicar
MARKDOWN_FILE = "articulo_completo.md"  # Archivo del artículo

print(f"📰 CONFIGURACIÓN DE PUBLICACIÓN:")
print(f"   📝 Título: {ARTICLE_TITLE}")
print(f"   📊 Estado: {PUBLICATION_STATUS}")
print(f"   📄 Archivo: {MARKDOWN_FILE}")
print(f"\n💡 Para cambiar el título, modifica ARTICLE_TITLE arriba")
print(f"💡 Para publicar directamente, cambia PUBLICATION_STATUS a 'publish'")
```

```python
print("📰 PUBLICANDO EN WORDPRESS...")
print("="*50)

# Verificar que el archivo existe
import os
if not os.path.exists(MARKDOWN_FILE):
    print(f"❌ Error: Archivo {MARKDOWN_FILE} no encontrado")
    print("   Ejecuta primero la celda de generación de contenido")
else:
    try:
        # Publicar en WordPress
        success = publish_article_from_markdown_cleaned(
            article_title=ARTICLE_TITLE,
            markdown_file_path=MARKDOWN_FILE,
            status=PUBLICATION_STATUS
        )
        
        if success:
            print(f"\n🎉 ¡ARTÍCULO PUBLICADO EXITOSAMENTE!")
            if PUBLICATION_STATUS == "draft":
                print(f"📝 Estado: BORRADOR (puedes revisarlo en WordPress antes de publicar)")
            else:
                print(f"📰 Estado: PUBLICADO (ya está visible en tu sitio web)")
        else:
            print(f"❌ Error al publicar el artículo")
            
    except Exception as e:
        print(f"❌ Error en la publicación: {str(e)}")
        import traceback
        traceback.print_exc()
```
