```python
import os
import pandas as pd
from dotenv import load_dotenv, find_dotenv
_ = load_dotenv(find_dotenv())

import google.generativeai as genai

# Configurar la API de Gemini (asumiendo que ya se ha configurado anteriormente)
genai.configure(api_key=os.environ['GEMINI_API_KEY'])

# Cargar el modelo Gemini 1.5 Flash
model = genai.GenerativeModel("gemini-1.5-flash")
```

```python
keywords = pd.read_csv("kr_af.csv")
```

```python
def seo_analisis_con_gemini(keywords):
    prompt = f""" Eres un experto en SEO y análisis de palabras clave. Tu tarea es analizar los siguientes datos de keyword research y realizar estas acciones:

    1. Clusterizar las palabras clave en las siguientes categorías:
       - Transaccionales
       - Informativas
       - Preguntas de usuarios
       - Otras

    2. Ordenar las palabras clave semánticamente según su relevancia y volumen.

    Datos de keywords:
    {keywords}

    Por favor, sigue estos pasos:

    1. Analiza cada palabra clave y asígnala a una de las categorías mencionadas.
    2. Dentro de cada categoría, ordena las palabras clave según su relevancia semántica y volumen de búsqueda.
    3. Presenta los resultados de manera clara y estructurada.

    Asegúrate de proporcionar una explicación breve de tu proceso de categorización y ordenamiento.

    Por favor, procesa estos datos y muestra los resultados."""
    
    try:
        response = model.generate_content(prompt)
        outline = response.text
        print("Analisis generado con éxito.")
        return outline
    except Exception as e:
        print(f"Error al generar el analisis con Gemini: {str(e)}")
        return "Error al generar el analisis con Gemini"

# Llamar a la función para generar el esquema
analisis = seo_analisis_con_gemini(keywords)

# Guardar el esquema en un archivo markdown
with open("analisis_keyword.md", "w", encoding="utf-8") as f:
    f.write(analisis)

print("\nAnalisis de keyword guardado en 'analisis_keyword.md'")


```
